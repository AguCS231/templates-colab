````python
%%writefile README.md
# templates-colab

**Plantillas reutilizables para proyectos de IA y Big Data en Google Colab + GitHub.**

Repositorio: [AguCS231/templates-colab](https://github.com/AguCS231/templates-colab)

---

## ¿Qué es esto?

Una colección de cuadernos y configuraciones listas para reutilizar en cualquier proyecto que combine **Google Colab** y **GitHub**.

Cada vez que empieces un proyecto nuevo, clonas este repositorio y copias la plantilla que necesites.

---

## ¿Por qué existe este repositorio?

Cuando trabajas con Colab, el sistema se reinicia continuamente y se pierde:
- El repositorio clonado.
- La configuración de Git.
- Las librerías instaladas.
- El token de GitHub.

Este repositorio contiene **el cuaderno que restaura todo eso** en 2-4 minutos. También contiene plantillas para las fases típicas de un proyecto de IA (EDA, modelado, evaluación, etc.).

---

## Contenido
````
