**Modifica solo la Celda 2** con los datos del proyecto. El resto se ejecuta tal cual.

**Antes de subir este cuaderno a GitHub**: borrar todas las salidas (`Edit → Clear all outputs`).

```python
# ============================================
# CONFIGURACIÓN DEL PROYECTO
# ============================================
# Modifica SOLO estas 3 variables.
# ============================================

GITHUB_USER = "AguCS231"
GIT_EMAIL   = "srcasares2@gmail.com"
REPO_NAME   = "lol"

# ============================================
# Variables derivadas (no tocar)
# ============================================

GIT_NAME   = GITHUB_USER
REPO_URL   = f"https://github.com/{GITHUB_USER}/{REPO_NAME}.git"
REPO_LOCAL = f"/content/{REPO_NAME}"

print("=== Configuración del proyecto ===")
print(f"Usuario GitHub:  {GITHUB_USER}")
print(f"Email Git:       {GIT_EMAIL}")
print(f"Repositorio:     {REPO_NAME}")
print(f"URL de clonado:  {REPO_URL}")
print(f"Carpeta local:   {REPO_LOCAL}")
```

```python
# ============================================
# Clonar repo y configurar Git
# ============================================
import os

if os.path.exists(REPO_LOCAL):
    print(f"Borrando {REPO_LOCAL} existente...")
    !rm -rf {REPO_LOCAL}

%cd /content
!git clone {REPO_URL}
%cd {REPO_LOCAL}

!git config --global user.email "{GIT_EMAIL}"
!git config --global user.name "{GIT_NAME}"

print("\n=== Verificación ===")
!pwd
!git config --list | grep user
```

```python
# ============================================
# Crear estructura de carpetas base
# ============================================
# Solo crea las que no existen. No borra contenido.
# ============================================

!mkdir -p docs
!mkdir -p notebooks
!mkdir -p src/data
!mkdir -p src/models
!mkdir -p src/security
!mkdir -p app
!mkdir -p tests
!mkdir -p data

# .gitkeep en las que estarán vacías al principio
!touch src/data/.gitkeep
!touch src/models/.gitkeep
!touch src/security/.gitkeep
!touch app/.gitkeep
!touch tests/.gitkeep
!touch data/.gitkeep

print("=== Estructura del proyecto ===")
!find . -type d -not -path './.git*' | sort
```

```python
# ============================================
# Capturar token de GitHub
# ============================================
# El token NO queda en la salida.
# Se pide cada vez que ejecutes la celda.
# ============================================
from getpass import getpass

TOKEN = getpass("Pega tu token de GitHub: ")
print(f"Token capturado: {len(TOKEN)} caracteres (esperado: 40)")
```

```python
# ============================================
# Configurar remote con token
# ============================================
# Verificación sin mostrar el token completo.
# ============================================

%cd {REPO_LOCAL}

!git remote set-url origin https://{TOKEN}@github.com/{GITHUB_USER}/{REPO_NAME}.git

# Verificar SIN exponer el token
!git remote get-url origin | grep -q "ghp_" && echo "✅ Token configurado correctamente"
```

```python
# ============================================
# Verificar estado del repo
# ============================================

print("=== Estado de Git ===")
!git status

print("\n=== Últimos commits ===")
!git log --oneline -5
```

```python
# ============================================
# Commit
# ============================================
# Si no hay cambios, este comando lo dirá. No pasa nada.
# ============================================

!git add .
!git commit -m "Estructura inicial del proyecto"
```

```python
# ============================================
# Push
# ============================================
# Si este comando falla, se puede reintentar
# ejecutando solo esta celda.
# ============================================

!git push
```
